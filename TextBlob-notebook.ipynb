<a id="sentiment-analysis"></a>
## 4. Sentiment Analysis

We love sentiment analysis!

The `.sentiment` property returns `Sentiment(polarity, subjectivity)`. Polarity ranges from -1.0 to 1.0, while subjectivity ranges from 0.0 to 1.0.

```python
testimonial = TextBlob("Textblob is amazingly simple to use. What great fun!")
```

```python
testimonial.sentiment
```

```python
testimonial.sentiment.polarity
```

<a id="tokenization"></a>
## 5. Tokenization

TextBlobs can be broken into words or sentences, tokenizing/chunking bigger strings. It's very ehlpful that TextBlob has built-in functions to chunk by word and by sentence.

```python
zen = TextBlob(
    "Beautiful is better than ugly. "
    "Explicit is better than implicit. "
    "Simple is better than complex."
)
```

```python
zen.words
```

```python
zen.sentences
```

```python
for sentence in zen.sentences:
    print(sentence.sentiment)
```

<a id="words-inflection-and-lemmatization"></a>
## 6. Words Inflection and Lemmatization

Each word in `TextBlob.words` is a `Word` object with useful methods for inflection.

```python
sentence = TextBlob("Use 4 spaces per indentation level.")
```

```python
sentence.words
```

```python
sentence.words[2].singularize()
```

```python
sentence.words[-1].pluralize()
```

Words can be lemmatized with `.lemmatize()`. (So smart! How???)

```python
w = Word("octopi")
```

```python
w.lemmatize()
```

```python
w = Word("went")
```

```python
w.lemmatize("v")
```

<a id="wordnet-integration"></a>
## 7. WordNet Integration

`Word` objects can access WordNet synsets and definitions.

```python
from textblob.wordnet import VERB
from textblob import Word
```

```python
word = Word("octopus")
```

```python
word.synsets
```

```python
Word("hack").get_synsets(pos=VERB)
```

```python
Word('octopus').definitions
```

```python
from textblob.wordnet import Synset

octopus = Synset("octopus.n.02")
shrimp = Synset("shrimp.n.03")
```

```python
octopus.path_similarity(shrimp)
```

<a id="spelling-correction"></a>
## 9. Spelling Correction

Use `.correct()` to attempt spelling correction.

```python
b = TextBlob("I havv goood speling!")
```

```python
print(b.correct())
```

```python
from textblob import Word
w = Word("falibility")
```

```python
w.spellcheck()
```

Spelling correction is useful for experimentation, but it's not guaranteed to be accurate (like any spellchecker/autocorrect), so it should be used carefully on real data.

<a id="word-and-noun-phrase-frequencies"></a>
## 10. Word and Noun Phrase Frequencies

TextBlob provides both `word_counts` and `.count()` for frequency information.

```python
monty = TextBlob(
    "We are no longer the Knights who say Ni. "
    "We are now the Knights who say Ekki ekki ekki PTANG."
)
```

```python
monty.word_counts['ekki']
```

```python
monty.words.count('ekki')
```

```python
monty.words.count('ekki', case_sensitive=True)
```

```python
wiki.noun_phrases.count('python')
```

<a id="textblobs-are-like-python-strings"></a>
## 12. TextBlobs Are Like Python Strings

TextBlobs support substring syntax, string methods, comparisons, and concatenation.

```python
zen[0:19]
```

```python
zen.upper()
```

```python
zen.find("Simple")
```

```python
apple_blob = TextBlob("apples")
banana_blob = TextBlob("bananas")
```

```python
apple_blob < banana_blob
```

```python
apple_blob == "apples"
```

```python
apple_blob + " and " + banana_blob
```

```python
"{0} and {1}".format(apple_blob, banana_blob)
```

## 13. n-grams

The `.ngrams()` method returns successive groups of `n` words.

```python
blob.ngrams(n=3)
```

<a id="get-start-and-end-indices-of-sentences"></a>
## 14. Get Start and End Indices of Sentences

Each sentence has `.start` and `.end` attributes showing where the sentence occurs within the original TextBlob.

```python
for s in zen.sentences:
    print(s)
    print("---- Starts at index {}, Ends at index {}".format(s.start, s.end))
```
